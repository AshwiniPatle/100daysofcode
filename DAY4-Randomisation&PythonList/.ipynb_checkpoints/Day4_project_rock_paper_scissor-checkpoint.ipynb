You are going to build a Rock, Paper, Scissors game. You will need to use what you have learnt about randomisation and Lists to achieve this.

Demo

https://appbrewery.github.io/python-day4-demo/

```python
rock = '''
    _______
---'   ____)
      (_____)
      (_____)
      (____)
---.__(___)
'''

paper = '''
    _______
---'   ____)____
          ______)
          _______)
         _______)
---.__________)
'''

scissors = '''
    _______
---'   ____)____
          ______)
       __________)
      (____)
---.__(___)
'''
import random
option = [rock, paper, scissors]
user_input = option[int(input("What do you choose? Type 0 for Rock, 1 for Paper and 2 for Scissors\n"))]
#print(option[user_input])
print(user_input)
print("Computer_choice:")
computer_input = random.choice(option)
print(computer_input)
if user_input == rock and computer_input == scissors:
    print("you win!")
elif user_input == scissors and computer_input == paper:
    print("you win!")
elif user_input == paper and computer_input == rock:
    print("you win!")
elif user_input== computer_input:
    print("its a toss")
else:
    print("You lose")
```
