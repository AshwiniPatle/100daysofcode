lets generate random floating point number

for this we will use random.random() - this only generates floating point number beyween 0 and 1 (includes 0 but not 1)


```python
random_number_0_to_1 = random.random()
print(random_number_0_to_1)
```

now what if I want something between 0 to 10

```python
random_number_0_to_1 = random.random() * 10
print(random_number_0_to_1)
```

another way of printing random number between 2 interval is to use random.uniform() which prints number between a and b inclusive of both. 

```python
floating_number = random.uniform(3, 4)
print(floating_number)
```

Create a coin flip program using what you have learnt about randomisation in Python. It should randomly print "Heads" or "Tails" everytime it is run.

```python
random_int = random.randint(0, 1)
if random_int == 0:
    print("Heads")
else:
    print("Tails")
```

Figure out how to pick a random name from the list of friends.

```python
friends = ["Alice", "Bob", "Charlie", "David", "Emanuel"]
random_int = random.randint(0,4)
print(friends[random_int])
```

there is another way to do it which is very simple

```python
friends = ["Alice", "Bob", "Charlie", "David", "Emanuel"]
print(random.choice(friends))
```
